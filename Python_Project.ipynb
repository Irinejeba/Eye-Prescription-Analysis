## **EYE PRESCRIPTION ANALYSIS**

```python
import os
import csv
import pandas as pd
import matplotlib.pyplot as plt

path = r"C:\Users\Admin\Desktop\CS\Advanced Python"

# ===== ADD AT TOP - Take 100 contents =====
df = pd.read_csv(os.path.join(path, "eye_prescription.csv"))
df_100 = df.head(100)  # Take 100 contents

print(f"Taking 100 contents from {len(df)} rows -> {len(df_100)} rows")
print(df_100.head())

# ===== Frequency  =====
def frequency(mylist):
    d = {}
    for x in mylist:
        d[x] = d.get(x, 0) + 1
    return d

right_freq = frequency(df_100['Right_SP'].astype(str).tolist())

# ===== new_csv =====
with open(os.path.join(path, "new_csv.csv"), "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["Sno", "Right_SP", "Patient_Count"])
    Sno = 1
    for k,v in right_freq.items(): #k: Eye power, v:Count
        w.writerow([Sno, k, v])
        Sno += 1

print("new_csv.csv created with 100 contents")

# ===== Chart =====

# 1. Right_SP 
right_counts = df_100['Right_SP'].value_counts().head(12)

plt.figure(figsize=(12,5))
plt.bar(right_counts.index.astype(str), right_counts.values, color='#5DA5DA', width=0.6)
plt.title("Right_SP - 100 Contents")
plt.xlabel("Right_SP")
plt.ylabel("Patient Count")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Age - HISTOGRAM - No overlapping 100%
plt.figure(figsize=(12,5))
plt.hist(df_100['Age'], bins=15, color='orange', edgecolor='black')
plt.title("Age - 100 Contents (Histogram)")
plt.xlabel("Age")
plt.ylabel("Patient Count")
plt.tight_layout()
plt.show()
```
